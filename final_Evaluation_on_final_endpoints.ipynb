```python
def format_docs(docs):
    # Sort by relevance score if available
    sorted_docs = sorted(docs, key=lambda x: x.metadata.get("score", 0), reverse=True)[:3]
    return "\n\n".join(doc.page_content for doc in sorted_docs)

@lru_cache()
def _load_classifier():
    return pipeline(
        "text-classification",
        model="grahamaco/question-complexity-classifier",
        top_k=1,
    )

# heuristic rules (fast)
_SPECIFIC_HOW_SIMPLE = {
    "how many", "how much", "how long", "how large",
    "how big", "how fast", "how small"
}
_SIMPLE_WH = {"who", "what", "when", "where", "which"}
_COMPLEX_IMP = {
    "describe", "explain", "discuss", "compare", "contrast",
    "analyze", "summarize", "elaborate", "define", "list", "tell", "tell me"
}

def _heuristic_label(q: str) -> str | None:
    q = q.strip().lower()
    words = q.split()
    if not words:
        return None
    first, first2 = words[0], " ".join(words[:2])
    if first2 in _SPECIFIC_HOW_SIMPLE or (len(words) <= 6 and first in _SIMPLE_WH):
        return "simple"
    if first in _COMPLEX_IMP or first in {"why", "how"}:
        return "complex"
    return None

def classify_question(question: str) -> str:                    #   ### CHANGED ###
    # 1) heuristics
    h = _heuristic_label(question)
    if h:
        return h
    # 2) DistilBERT fallback
    result = _load_classifier()(question.strip())[0][0]["label"]  # note [0][0]
    return "simple" if result.upper() == "SIMPLE" else "complex"


def get_retriever():
    embedder = HuggingFaceEmbeddings(model_name=EMBEDDING_MODEL_NAME)
    client = QdrantClient(url=QDRANT_URL, api_key=QDRANT_API_KEY, prefer_grpc=True)
    if client.get_collection(COLLECTION_NAME).vectors_count == 0:
        raise RuntimeError("Qdrant collection is empty.")

    vectorstore = QdrantVectorStore(
        client=client,
        collection_name=COLLECTION_NAME,
        embeddings=embedder
    )

    return vectorstore
```

```python
from transformers import AutoModelForCausalLM, AutoTokenizer, pipeline
from langchain.llms.base import LLM
from typing import List
from pydantic import PrivateAttr
from langchain.prompts import ChatPromptTemplate
from langchain.schema.runnable import RunnableLambda, RunnablePassthrough

# Step 1: Load Qwin model from Hugging Face
def load_qwin_pipeline():
    model_id = "Qwen/Qwen2-0.5B"

    model = AutoModelForCausalLM.from_pretrained(
        model_id,
        device_map="auto",
        trust_remote_code=True  # Important for Qwen models
    )
    tokenizer = AutoTokenizer.from_pretrained(
        model_id,
        trust_remote_code=True  # Required for Qwen tokenizer to work correctly
    )

    pipe = pipeline("text-generation", model=model, tokenizer=tokenizer)
    return pipe


# Step 2: LangChain-compatible wrapper
class QwinChatLLM(LLM):
    _pipeline: any = PrivateAttr()

    def __init__(self, pipeline):
        super().__init__()
        self._pipeline = pipeline

    def _call(self, prompt: str, stop: List[str] = None) -> str:
        output = self._pipeline(prompt, max_new_tokens=200, do_sample=False)[0]["generated_text"]
        return output.replace(prompt, "").strip()

    @property
    def _llm_type(self) -> str:
        return "qwin-chat"

def build_onchain_with_qwin_model():
    qwin_pipeline = load_qwin_pipeline()

    # Use same model for both complexity levels
    llm_simple = QwinChatLLM(qwin_pipeline)
    llm_complex = QwinChatLLM(qwin_pipeline)
    llm_by_label = {"simple": llm_simple, "complex": llm_complex}

    retriever = get_retriever().as_retriever(search_kwargs={"k": 5})

    # Prompt with few-shot examples
    prompt_template = """
You are Ramesses II, Pharaoh of Egypt. Speak as yourself.

################## INSTRUCTIONS ##################
- Use **only** the provided context.
- If the answer is not in the context, say: “I do not know.”
- Be direct, clear, and avoid invented stories.
- If the question is simple, reply briefly (≤ 3 sentences).
- If the question is complex, explain in more depth (≤ 6 sentences).
- Do not repeat the question.

################## EXAMPLES ##################

### Example 1 (Simple):
Q: How old were you when you became Pharaoh?
A: I was crowned Pharaoh at the age of 14.

### Example 2 (Complex):
Q: Why was the Battle of Kadesh important to your reign?
A: The Battle of Kadesh was one of my greatest campaigns. Though inconclusive, it showed my strength as a military leader and led to the world’s first known peace treaty with the Hittites.

───────────────────────────────────────────────

### Context:
{context}

### Visitor’s Question:
{input}

### Your Answer:
"""
    chat_prompt = ChatPromptTemplate.from_template(prompt_template)

    # Your existing classifier (complex/simple)
    classify_node = RunnableLambda(classify_question)

    # Format context (injected from retriever)
    def format_docs(docs):
        sorted_docs = sorted(docs, key=lambda x: x.metadata.get("score", 0), reverse=True)[:5]
        return "\n\n---\n\n".join(doc.page_content for doc in sorted_docs)

    # Your grounding similarity checker
    def check_grounding(answer: str, context_list: list[str], threshold=0.55) -> bool:
        from langchain.embeddings import HuggingFaceEmbeddings
        import numpy as np

        embedding_model = HuggingFaceEmbeddings(model_name="sentence-transformers/all-mpnet-base-v2")
        answer_emb = embedding_model.embed_query(answer)
        context_embs = [embedding_model.embed_query(ctx) for ctx in context_list]

        similarities = [
            np.dot(answer_emb, ctx_emb) / (np.linalg.norm(answer_emb) * np.linalg.norm(ctx_emb))
            for ctx_emb in context_embs
        ]

        return max(similarities, default=0) >= threshold

    # Build payload
    gather = {
        "context": retriever | format_docs,
        "input": RunnablePassthrough(),
        "complexity": classify_node,
    }

    def build_payload(d):
        return {
            "prompt": chat_prompt.format_prompt(**d),
            "complexity": d["complexity"],
            "question": d["input"],  # used in call_llm for grounding
        }

    def call_llm(d):
        llm = llm_by_label[d["complexity"]]
        response = llm.invoke(d["prompt"].to_messages())
        answer = response.content if hasattr(response, "content") else str(response)

        # Grounding check
        raw_contexts = retriever.invoke(d["question"])
        raw_contexts = [doc.page_content for doc in raw_contexts]

        if not check_grounding(answer, raw_contexts):
            print(f"Unfaithful answer rejected:\nQ: {d['question']}\nA: {answer}\n")
            return "I do not know. The answer is not present in the given context."

        return answer
        return answer

    # Final RAG chain
    chain = gather | RunnableLambda(build_payload) | RunnableLambda(call_llm)

    def wrapped(user_input: str) -> str:
        return chain.invoke(user_input)

    return wrapped

query = "Talk about your wife"
chain = build_onchain_with_qwin_model()
# Process the query through the RAG pipeline
response = chain(query)
print(f"Response: {response}")
```

```python

def build_onchain_with_qwin_model():
    qwin_pipeline = load_qwin_pipeline()

    # Use same model for both complexity levels
    llm_simple = QwinChatLLM(qwin_pipeline)
    llm_complex = QwinChatLLM(qwin_pipeline)
    llm_by_label = {"simple": llm_simple, "complex": llm_complex}

    retriever = get_retriever().as_retriever(search_kwargs={"k": 5})

    # Prompt with few-shot examples
    prompt_template = """
You are Ramesses II, Pharaoh of Egypt. Speak as yourself.

################## INSTRUCTIONS ##################
- Use **only** the provided context.
- If the answer is not in the context, say: “I do not know.”
- Be direct, clear, and avoid invented stories.
- If the question is simple, reply briefly (≤ 3 sentences).
- If the question is complex, explain in more depth (≤ 6 sentences).
- Do not repeat the question.

################## EXAMPLES ##################

### Example 1 (Simple):
Q: How old were you when you became Pharaoh?
A: I was crowned Pharaoh at the age of 14.

### Example 2 (Complex):
Q: Why was the Battle of Kadesh important to your reign?
A: The Battle of Kadesh was one of my greatest campaigns. Though inconclusive, it showed my strength as a military leader and led to the world’s first known peace treaty with the Hittites.

───────────────────────────────────────────────

### Context:
{context}

### Visitor’s Question:
{input}

### Your Answer:
"""
    chat_prompt = ChatPromptTemplate.from_template(prompt_template)

    # Your existing classifier (complex/simple)
    classify_node = RunnableLambda(classify_question)

    # Format context (injected from retriever)
    def format_docs(docs):
        sorted_docs = sorted(docs, key=lambda x: x.metadata.get("score", 0), reverse=True)[:5]
        return "\n\n---\n\n".join(doc.page_content for doc in sorted_docs)

    # Your grounding similarity checker
    def check_grounding(answer: str, context_list: list[str], threshold=0.55) -> bool:
        from langchain.embeddings import HuggingFaceEmbeddings
        import numpy as np

        embedding_model = HuggingFaceEmbeddings(model_name="sentence-transformers/all-mpnet-base-v2")
        answer_emb = embedding_model.embed_query(answer)
        context_embs = [embedding_model.embed_query(ctx) for ctx in context_list]

        similarities = [
            np.dot(answer_emb, ctx_emb) / (np.linalg.norm(answer_emb) * np.linalg.norm(ctx_emb))
            for ctx_emb in context_embs
        ]

        return max(similarities, default=0) >= threshold

    # Build payload
    gather = {
        "context": retriever | format_docs,
        "input": RunnablePassthrough(),
        "complexity": classify_node,
    }

    def build_payload(d):
        return {
            "prompt": chat_prompt.format_prompt(**d),
            "complexity": d["complexity"],
            "question": d["input"],  # used in call_llm for grounding
        }

    def call_llm(d):
        llm = llm_by_label[d["complexity"]]
        response = llm.invoke(d["prompt"].to_messages())
        answer = response.content if hasattr(response, "content") else str(response)

        # Grounding check
        raw_contexts = retriever.invoke(d["question"])
        raw_contexts = [doc.page_content for doc in raw_contexts]

        if not check_grounding(answer, raw_contexts):
            print(f"Unfaithful answer rejected:\nQ: {d['question']}\nA: {answer}\n")
            return "I do not know. The answer is not present in the given context."

        return answer
        return answer

    # Final RAG chain
    chain = gather | RunnableLambda(build_payload) | RunnableLambda(call_llm)

    def wrapped(user_input: str) -> str:
        return chain.invoke(user_input)

    return wrapped


def save_evaluation_data(evaluation_data, filepath="evaluation_data.csv"):
    df = pd.DataFrame(evaluation_data)
    df.to_csv(filepath, index=False)
    print(f"Evaluation dataset saved to {filepath}")

def load_evaluation_data(filepath="evaluation_data.csv"):
    if os.path.exists(filepath):
        df = pd.read_csv(filepath, encoding="ISO-8859-1")
        df["contexts"] = df["contexts"].apply(json.loads)
        return df.to_dict(orient="records")
    else:
        raise FileNotFoundError(f"No dataset found at {filepath}")

def create_test_dataset(test_size=10, save_path="evaluation_data.csv"):
    """
    Generate a synthetic test dataset for RAG evaluation using RAGAS.

    Args:
        test_size (int): Number of question-answer pairs to generate.
        save_path (str): Path to save the generated dataset.

    Returns:
        list: List of dictionaries containing questions, ground truths, contexts, and answers.
    """

    llm = ChatOpenAI(
        model="gpt-3.5-turbo",
        openai_api_key=OPENAI_API_KEY,
        max_tokens=1024  # or higher
    )

    embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-mpnet-base-v2")

    vectorstore = get_retriever()

    # Qdrant documents with multiple queries
    seed_queries = ["Ramesses II", "Battle of Kadesh", "Hittite Treaty", "Abu Simbel"]
    documents = []
    retriever = vectorstore.as_retriever(search_kwargs={"k": 5})
    for query in seed_queries:
        try:
            docs = retriever.invoke(query)
            documents.extend(
                Document(page_content=doc.page_content, metadata=doc.metadata)
                for doc in docs
            )
        except Exception as e:
            logging.error(f"Failed to retrieve documents for query '{query}': {str(e)}")

    # Deduplicate documents
    documents = list({doc.page_content: doc for doc in documents}.values())

    generator = TestsetGenerator.from_langchain(
        llm=llm,
        embedding_model=embeddings,
    )

    testset = generator.generate_with_langchain_docs(
        documents=documents,
        testset_size=test_size
    )

    evaluation_data = testset.to_pandas()

    # Rename columns to match RAGAs expectations
    evaluation_data = evaluation_data.rename(columns={
        "user_input": "question",
        "reference_contexts": "contexts",
        "reference": "ground_truth"
    })

    if "answer" not in evaluation_data.columns:
            evaluation_data["answer"] = [None for _ in range(len(evaluation_data))]

    ########### change here
    chain = build_onchain_with_qwin_model()
    vector_retriever = vectorstore

    for idx, item in evaluation_data.iterrows():
        retrieved_docs = vector_retriever.as_retriever(search_kwargs={"k": 5}).invoke(item["question"])
        evaluation_data.loc[idx, "contexts"] =  json.dumps([doc.page_content for doc in retrieved_docs])
        evaluation_data.loc[idx, "answer"] = chain(item["question"])
        time.sleep(2)  # Light throttling for OpenAI

    evaluation_data = evaluation_data[["question", "answer", "contexts", "ground_truth"]].to_dict(orient="records")
    save_evaluation_data(evaluation_data, save_path)

    return evaluation_data

def evaluate_rag_model(filepath="evaluation_data.csv"):
    """
    Evaluate the RAG model using RAGAS metrics.

    Args:
        evaluation_data (list, optional): List of dictionaries with questions, ground truths, contexts, and answers.
        filepath (str): Path to load the saved dataset if evaluation_data is None.

    Returns:
        dict: RAGAS evaluation results.
    """
    evaluation_data = load_evaluation_data(filepath)

    llm = ChatOpenAI(
        model="gpt-3.5-turbo",
        openai_api_key=OPENAI_API_KEY,
        max_tokens=1024  # or higher
    )

    embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-mpnet-base-v2")

    ragas_dataset = Dataset.from_dict({
        "question": [item["question"] for item in evaluation_data],
        "answer": [item["answer"] for item in evaluation_data],
        "contexts": [json.loads(item["contexts"]) if isinstance(item["contexts"], str) else item["contexts"] for item in evaluation_data],
        "ground_truth": [item["ground_truth"] for item in evaluation_data]
    })

    metrics = [
        Faithfulness(),
        AnswerRelevancy(),
        ContextPrecision(),
        ContextRecall(),
        NoiseSensitivity(),
        ContextEntityRecall()
        ]

    results = evaluate(
        dataset=ragas_dataset,
        metrics=metrics,
        llm=llm,
        embeddings=embeddings
    )

    return results

# Example usage
if __name__ == "__main__":
    evaluation_data = create_test_dataset(test_size=10, save_path="Qwen_eval.csv")
    print("Evaluation Data:", evaluation_data)
    df = pd.read_csv("Qwen_eval.csv")
    print(df.head(10))
    results = evaluate_rag_model(filepath="Qwen_eval.csv")
    print("Evaluation Results:", results)
```

```python
import numpy as np

def check_grounding(answer: str, context_list: list[str], threshold=0.5) -> bool:
    """
    Check if the generated answer is semantically similar to any of the retrieved context passages.

    Args:
        answer (str): The LLM's response.
        context_list (list[str]): Retrieved documents.
        threshold (float): Cosine similarity threshold to accept grounding.

    Returns:
        bool: True if the answer is grounded; False if likely hallucinated.
    """
    embedding_model = HuggingFaceEmbeddings(model_name="sentence-transformers/all-mpnet-base-v2")

    answer_emb = embedding_model.embed_query(answer)
    context_embs = [embedding_model.embed_query(ctx) for ctx in context_list]

    similarities = [
        np.dot(answer_emb, ctx_emb) / (np.linalg.norm(answer_emb) * np.linalg.norm(ctx_emb))
        for ctx_emb in context_embs
    ]

    max_sim = max(similarities, default=0)
    return max_sim >= threshold

def format_docs(docs):
    # Sort by relevance score if available
    sorted_docs = sorted(docs, key=lambda x: x.metadata.get("score", 0), reverse=True)[:5]
    return "\n\n---\n\n".join(doc.page_content for doc in sorted_docs)

def build_onchain_without_multiretrieval():
    # Set up LLMs with slightly increased token limits
    llm_simple = ChatGoogleGenerativeAI(
        model="gemini-1.5-flash",
        google_api_key=GOOGLE_API_KEY,
        max_output_tokens=80,  # Increased from 60
    )
    llm_complex = ChatGoogleGenerativeAI(
        model="gemini-1.5-flash",
        google_api_key=GOOGLE_API_KEY,
        max_output_tokens=150,  # Reduced from 200 for conciseness
    )
    llm_by_label = {"simple": llm_simple, "complex": llm_complex}

    # Set up retriever
    vector_retriever = get_retriever()
    retriever = vector_retriever.as_retriever(search_kwargs={"k": 5})

    # Grounded, clarity-first prompt
    prompt_template = """
      You are Ramesses II, Pharaoh of Egypt. Speak as yourself.

      ################## INSTRUCTIONS ##################
      - Use **only** the provided context.
      - If the answer is not in the context, say: “I do not know.”
      - Be direct, clear, and avoid invented stories.
      - If the question is simple, reply briefly (≤ 2 sentences).
      - If the question is complex, explain in more depth (≤ 6 sentences).
      - Do not repeat the question.

      ################## EXAMPLES ##################

      ### Example 1 (Simple):
      Q: How old were you when you became Pharaoh?
      A: I was crowned Pharaoh at the age of 14.

      ### Example 2 (Simple):
      Q: What was Abu Simbel?
      A: Abu Simbel is a great temple I built to honor the gods and my legacy.

      ### Example 3 (Complex):
      Q: Why was the Battle of Kadesh important to your reign?
      A: The Battle of Kadesh was one of my greatest campaigns. Though inconclusive, it showed my strength as a military leader and led to the world’s first known peace treaty with the Hittites.

      ### Example 4 (Complex):
      Q: What were the causes of the Battle of Kadesh, and how did it end?
      A: The battle was sparked by territorial disputes over Syria between Egypt and the Hittites. It ended inconclusively, but led to a historic peace treaty between the two empires.

      ───────────────────────────────────────────────

      ### Context:
      {context}

      ### Visitor’s Question:
      {input}

      ### Your Answer:
      """


    chat_prompt = ChatPromptTemplate.from_template(prompt_template)

    classify_node = RunnableLambda(classify_question)

    gather = {
        "context": retriever | format_docs,
        "input": RunnablePassthrough(),
        "complexity": classify_node,
    }

    def build_payload(d):
        return {
        "prompt": chat_prompt.format_prompt(**d),
        "complexity": d["complexity"],
        "question": d["input"],  # Add this line
        }

    def call_llm(d):
        llm = llm_by_label[d["complexity"]]
        response = llm.invoke(d["prompt"].to_messages())
        answer = response.content if hasattr(response, "content") else str(response)

        # Re-run the retriever to get raw contexts (not formatted)
        raw_contexts = retriever.invoke(d["question"])
        raw_contexts = [doc.page_content for doc in raw_contexts]

        # Post-check grounding
        if not check_grounding(answer, raw_contexts):
            print(f"Unfaithful answer rejected:\nQ: {d['question']}\nA: {answer}\n")
            return "I do not know. The answer is not present in the given context."

        return answer


    # Compose the full chain
    chain = gather | RunnableLambda(build_payload) | RunnableLambda(call_llm)

    def wrapped(user_input: str) -> str:
        return chain.invoke(user_input)

    return wrapped

def save_evaluation_data(evaluation_data, filepath="evaluation_data.csv"):
    df = pd.DataFrame(evaluation_data)
    df.to_csv(filepath, index=False)
    print(f"Evaluation dataset saved to {filepath}")

def load_evaluation_data(filepath="evaluation_data.csv"):
    if os.path.exists(filepath):
        df = pd.read_csv(filepath, encoding="ISO-8859-1")
        df["contexts"] = df["contexts"].apply(json.loads)
        return df.to_dict(orient="records")
    else:
        raise FileNotFoundError(f"No dataset found at {filepath}")

def create_test_dataset(test_size=10, save_path="evaluation_data.csv"):
    """
    Generate a synthetic test dataset for RAG evaluation using RAGAS.

    Args:
        test_size (int): Number of question-answer pairs to generate.
        save_path (str): Path to save the generated dataset.

    Returns:
        list: List of dictionaries containing questions, ground truths, contexts, and answers.
    """

    llm = ChatOpenAI(
        model="gpt-3.5-turbo",
        openai_api_key=OPENAI_API_KEY,
        max_tokens=1024  # or higher
    )

    embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-mpnet-base-v2")

    vectorstore = get_retriever()

    # Qdrant documents with multiple queries
    seed_queries = ["Ramesses II", "Battle of Kadesh", "Hittite Treaty", "Abu Simbel"]
    documents = []
    retriever = vectorstore.as_retriever(search_kwargs={"k": 5})
    for query in seed_queries:
        try:
            docs = retriever.invoke(query)
            documents.extend(
                Document(page_content=doc.page_content, metadata=doc.metadata)
                for doc in docs
            )
        except Exception as e:
            logging.error(f"Failed to retrieve documents for query '{query}': {str(e)}")

    # Deduplicate documents
    documents = list({doc.page_content: doc for doc in documents}.values())

    generator = TestsetGenerator.from_langchain(
        llm=llm,
        embedding_model=embeddings,
    )

    testset = generator.generate_with_langchain_docs(
        documents=documents,
        testset_size=test_size
    )

    evaluation_data = testset.to_pandas()

    # Rename columns to match RAGAs expectations
    evaluation_data = evaluation_data.rename(columns={
        "user_input": "question",
        "reference_contexts": "contexts",
        "reference": "ground_truth"
    })

    if "answer" not in evaluation_data.columns:
            evaluation_data["answer"] = [None for _ in range(len(evaluation_data))]

    ########### change here
    chain = build_onchain_without_multiretrieval()
    vector_retriever = vectorstore

    for idx, item in evaluation_data.iterrows():
        retrieved_docs = vector_retriever.as_retriever(search_kwargs={"k": 5}).invoke(item["question"])
        evaluation_data.loc[idx, "contexts"] =  json.dumps([doc.page_content for doc in retrieved_docs])
        evaluation_data.loc[idx, "answer"] = chain(item["question"])
        time.sleep(2)  # Light throttling for OpenAI

    evaluation_data = evaluation_data[["question", "answer", "contexts", "ground_truth"]].to_dict(orient="records")
    save_evaluation_data(evaluation_data, save_path)

    return evaluation_data

def evaluate_rag_model(filepath="evaluation_data.csv"):
    """
    Evaluate the RAG model using RAGAS metrics.

    Args:
        evaluation_data (list, optional): List of dictionaries with questions, ground truths, contexts, and answers.
        filepath (str): Path to load the saved dataset if evaluation_data is None.

    Returns:
        dict: RAGAS evaluation results.
    """
    evaluation_data = load_evaluation_data(filepath)

    llm = ChatOpenAI(
        model="gpt-3.5-turbo",
        openai_api_key=OPENAI_API_KEY,
        max_tokens=1024  # or higher
    )

    embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-mpnet-base-v2")

    ragas_dataset = Dataset.from_dict({
        "question": [item["question"] for item in evaluation_data],
        "answer": [item["answer"] for item in evaluation_data],
        "contexts": [json.loads(item["contexts"]) if isinstance(item["contexts"], str) else item["contexts"] for item in evaluation_data],
        "ground_truth": [item["ground_truth"] for item in evaluation_data]
    })

    metrics = [
        Faithfulness(),
        AnswerRelevancy(),
        ContextPrecision(),
        ContextRecall(),
        ]

    results = evaluate(
        dataset=ragas_dataset,
        metrics=metrics,
        llm=llm,
        embeddings=embeddings
    )

    return results

import pandas as pd

# Example usage
if __name__ == "__main__":
    evaluation_data = create_test_dataset(test_size=10)
    print("Evaluation Data:", evaluation_data)
    df = pd.read_csv("evaluation_data.csv")
    print(df.head(10))
    results = evaluate_rag_model()
    df = pd.DataFrame([results])
    print(df)
```
