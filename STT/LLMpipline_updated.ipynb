```python
def process_audio(audio_file_name, prompt):
  # print("Uploading file...")
  audio_file = genai.upload_file(path=audio_file_name)

  # print("Making LLM inference request...")
  response = model.generate_content(
      [audio_file, prompt],
      request_options={"timeout": 600}
  )

  # Print the response, rendering any Markdown
  # print(response.text)

  return response.text
```

```python
# chunking
def chunk_text(text, max_length=200):
    """Splits the text into smaller chunks of specified max length."""

    chunks = []
    words = text.split()
    current_chunk = ''
    for word in words:
        if len(current_chunk) + len(word) <= max_length:
            current_chunk += ' ' + word
        else:
            chunks.append(current_chunk.strip())
            current_chunk = word

    if current_chunk:
        chunks.append(current_chunk.strip())

    return chunks

async def async_tts_worker(text_chunks, cloning_audio):
    """Processes text chunks asynchronously for TTS streaming."""
    audio_urls = []

    first_chunk_time = None  # Track time for first chunk

    for i, chunk in enumerate(text_chunks):
        if i == 0:
          first_start = time.time()

        handler = await fal_client.submit_async(
            "fal-ai/zonos",
            arguments={
                "reference_audio_url": cloning_audio,
                "prompt": chunk
            },
            webhook_url="https://optional.webhook.url/for/results",
        )
        request_id = handler.request_id
        result = await fal_client.result_async("fal-ai/zonos", request_id)

        audio_url = result["audio"]["url"]
        audio_urls.append(audio_url)

        # Measure time to first audio chunk
        if i == 0:
          first_chunk_time = time.time() - first_start
          print(f"⏳ Time to first audio: {first_chunk_time:.2f} seconds")

        # Play audio as soon as it's ready
        play_audio(audio_url)

    return audio_urls

def play_audio(audio_url):
    """Plays the generated audio sequentially."""
    if audio_url:
        display(Audio(audio_url, autoplay=True))

async def process_and_play(audio_file_name, prompt):
    """Runs the full pipeline asynchronously with streaming."""
    # Process Audio -> Get Text
    response_text = process_audio(audio_file_name, prompt)
    # Chunk text
    text_chunks = chunk_text(response_text, max_length=200)

    # Upload reference audio
    cloning_audio = fal_client.upload_file(audio_file_name)

    # Start TTS processing and streaming
    start = time.time()
    await async_tts_worker(text_chunks, cloning_audio)
    end = time.time()

    print(f"Total Processing Time: {end - start:.2f} seconds")

# Define input file and prompt
audio_file_name = "/content/ALLTALK_MaleBritishAccent_NarratorKnightley_Voice_1.wav"
prompt = "Can you generate a 200 word essay based on this audio"

# Run the async processing pipeline
nest_asyncio.apply()
asyncio.run(process_and_play(audio_file_name, prompt))
```

```python
# threads
def play_audio_stream(audio_urls):
  """Plays the generated audio sequentially"""
  for audio_url in audio_urls:
      if audio_url:
          # print(f"Playing audio: {audio_url}")
          display(Audio(audio_url, autoplay=True))
          time.sleep(2)  # Small delay to avoid overlap

def tts_worker_thread(text_response, clonning_audio, result_list, index):
    """Handles TTS generation in a separate thread"""
    # print(f"Processing TTS for part {index+1}...")

    # Send request to TTS model
    handler = fal_client.submit(
        "fal-ai/zonos",
        arguments={
            "reference_audio_url": clonning_audio,
            "prompt": text_response
        },
        webhook_url="https://optional.webhook.url/for/results",
    )

    request_id = handler.request_id
    # print(f"Request ID: {request_id}")

    # Wait for processing to complete
    status = fal_client.status("fal-ai/zonos", request_id, with_logs=True)
    # print(f"Status: {status}")

    # Retrieve audio result
    result = fal_client.result("fal-ai/zonos", request_id)
    audio_url = result["audio"]["url"]

    # print(f"Generated Audio URL: {audio_url}")

    # Store the result in the shared list
    result_list[index] = audio_url
    # print(f"Result stored in index {index}")

def split_text_equally(text):
    """Splits the text into two nearly equal parts at the nearest '.'"""
    sentences = text.split(".")  # Split into sentences
    total_sentences = len(sentences)

    # Find the midpoint
    mid = total_sentences // 2

    # Ensure we don't split in the middle of a sentence
    part1 = ".".join(sentences[:mid]) + "." if mid > 0 else sentences[0]
    part2 = ".".join(sentences[mid:]) if mid < total_sentences else ""

    return part1.strip(), part2.strip()

def generate_audio_from_response_using_threads(response_text, clonning_audio):
    """Splits response, processes it in parallel threads, and returns audio results"""
    # Split text into two parts
    part1, part2 = split_text_equally(response_text)

    # Create a shared result list
    result_list = [None, None]

    # Create two threads for parallel TTS conversion
    thread1 = threading.Thread(target=tts_worker_thread, args=(part1, clonning_audio, result_list, 0))
    thread2 = threading.Thread(target=tts_worker_thread, args=(part2, clonning_audio, result_list, 1))

    # Start threads
    thread1.start()
    thread2.start()

    # Wait for both threads to complete
    thread1.join()
    thread2.join()

    return result_list
```

```python
# thread pool
import threading
import concurrent.futures

def tts_worker_threads(text_response, clonning_audio):
    """Handles TTS generation in a separate thread"""
    handler = fal_client.submit(
        "fal-ai/zonos",
        arguments={
            "reference_audio_url": clonning_audio,
            "prompt": text_response
        },
        webhook_url="https://optional.webhook.url/for/results",
    )

    request_id = handler.request_id

    status = fal_client.status("fal-ai/zonos", request_id, with_logs=True)

    result = fal_client.result("fal-ai/zonos", request_id)
    return result["audio"]["url"]  # Return the generated URL

def split_text_equally(text):
    """Splits the text into two nearly equal parts at the nearest '.'"""
    sentences = text.split(".")
    total_sentences = len(sentences)
    mid = total_sentences // 2

    part1 = ".".join(sentences[:mid]) + "." if mid > 0 else sentences[0]
    part2 = ".".join(sentences[mid:]) if mid < total_sentences else ""

    return part1.strip(), part2.strip()

def generate_audio_from_response_using_threads(response_text, clonning_audio):
    """Splits response, processes it in parallel threads, and returns audio results"""

    part1, part2 = split_text_equally(response_text)

    with concurrent.futures.ThreadPoolExecutor(max_workers=2) as executor:
        future1 = executor.submit(tts_worker_threads, part1, clonning_audio)
        future2 = executor.submit(tts_worker_threads, part2, clonning_audio)

        result_list = [future1.result(), future2.result()]  # Get results from both threads

    return result_list
```

```python
# multiprocessing
import multiprocessing

def tts_worker_mp_pool(args):
    """
    Handles TTS generation in a separate process using a Pool.
    Receives a tuple (text_response, clonning_audio) and returns the audio URL.
    """
    text_response, clonning_audio = args
    handler = fal_client.submit(
        "fal-ai/zonos",
        arguments={
            "reference_audio_url": clonning_audio,
            "prompt": text_response
        },
        webhook_url="https://optional.webhook.url/for/results",
    )

    request_id = handler.request_id
    status = fal_client.status("fal-ai/zonos", request_id, with_logs=True)
    result = fal_client.result("fal-ai/zonos", request_id)
    return result["audio"]["url"]

def split_text_equally(text):
    """Splits the text into two nearly equal parts at the nearest '.'"""
    sentences = text.split(".")
    total_sentences = len(sentences)
    mid = total_sentences // 2

    part1 = ".".join(sentences[:mid]) + "." if mid > 0 else sentences[0]
    part2 = ".".join(sentences[mid:]) if mid < total_sentences else ""

    return part1.strip(), part2.strip()

def generate_audio_from_response_using_pool(response_text, clonning_audio):
    """Splits response, processes it in parallel using a Pool, and returns audio results."""
    part1, part2 = split_text_equally(response_text)
    with multiprocessing.Pool(processes=2) as pool:
        # Map the function to the two tasks (each a tuple of (text, clonning_audio))
        result_list = pool.map(tts_worker_mp_pool, [(part1, clonning_audio), (part2, clonning_audio)])
    return result_list
```
