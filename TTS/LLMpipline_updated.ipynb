```python
dotenv.load_dotenv()

genai.configure(api_key=os.getenv("GOOGLE_API_KEY"))

# Set up the model
generation_config = {
    "temperature": 0.2,
    "top_p": 0.8,
    "top_k": 64,
    "max_output_tokens": 1024,
}
# Load the model
model = genai.GenerativeModel(
    model_name="models/gemini-1.5-flash-latest",
)

def process_audio(audio_file_name, prompt):
  """ func to request from large langauge model """

  audio_file = genai.upload_file(path=audio_file_name)
  response = model.generate_content(
      [audio_file, prompt],
      request_options={"timeout": 600}
  )

  return response.text
```

```python
def play_audio(audio_url):
    """Plays audio and blocks until finished"""
    audio = Audio(audio_url, autoplay=True)
    display(audio)
    duration = len(audio.data) / (44100 * 2)  # Sample rate * channels
    time.sleep(duration)  # Add small buffer


def chunk_text(text, max_length=200):
    """Splits the text into smaller chunks of specified max length."""

    sentences = re.split(r'(?<=[.!?])\s+', text)
    chunks = []
    current_chunk = ""

    for sentence in sentences:
        if len(current_chunk) + len(sentence) <= max_length:
            current_chunk += " " + sentence
        else:
            chunks.append(current_chunk.strip())
            current_chunk = sentence

    if current_chunk:
        chunks.append(current_chunk.strip())

    return chunks

async def async_tts_worker(text_chunks, cloning_audio):
    """Processes text chunks asynchronously for TTS streaming."""
    audio_urls = []

    first_chunk_time = None  # Track time for first chunk

    for i, chunk in enumerate(text_chunks):
        if i == 0:
          first_start = time.time()

        handler = await fal_client.submit_async(
            "fal-ai/zonos",
            arguments={
                "reference_audio_url": cloning_audio,
                "prompt": chunk
            },
            webhook_url="https://optional.webhook.url/for/results",
        )
        request_id = handler.request_id
        result = await fal_client.result_async("fal-ai/zonos", request_id)

        audio_url = result["audio"]["url"]
        audio_urls.append(audio_url)

        # Measure time to first audio chunk
        if i == 0:
          first_chunk_time = time.time() - first_start
          print(f"⏳ Time to first audio: {first_chunk_time:.2f} seconds")

        # Play audio as soon as it's ready
        play_audio(audio_url)

    return audio_urls

#------------------------------
# async def fetch_result(rid):
#     start = time.time()
#     result = await fal_client.result_async("fal-ai/zonos", rid)
#     print(f"Request {rid} took {time.time()-start:.2f}s")
#     return result

async def async_tts_worker(text_chunks, cloning_audio):
    """Processes text chunks in parallel for TTS streaming."""
    request_ids = []
    for chunk in text_chunks:
        handler = await fal_client.submit_async(
            "fal-ai/zonos",
            arguments={
                "reference_audio_url": cloning_audio,
                "prompt": chunk
            }
        )
        request_ids.append(handler.request_id)  # Get the ID after awaiting

    # Then await all results in parallel
    results = await asyncio.gather(*[
        fal_client.result_async("fal-ai/zonos", request_id)
        for request_id in request_ids
    ])
    results = await asyncio.gather(*[fetch_result(rid) for rid in request_ids])

    audio_urls = []
    for result in results:
        audio_url = result["audio"]["url"]
        audio_urls.append(audio_url)
        play_audio(audio_url)

    return audio_urls

#------------------------------
async def process_and_play(audio_file_name, prompt):
    """Runs the full pipeline asynchronously with streaming."""
    # Process Audio -> Get Text
    response_text = process_audio(audio_file_name, prompt)
    # Chunk text
    text_chunks = chunk_text(response_text, max_length=300)

    # Upload reference audio
    cloning_audio = fal_client.upload_file(audio_file_name)

    # Start TTS processing and streaming
    start = time.time()
    audio_urls = await async_tts_worker(text_chunks, cloning_audio)
    end = time.time()

    print(f"Total Processing Time: {end - start:.2f} seconds")

# Define input file and prompt
audio_file_name = "/content/ALLTALK_MaleBritishAccent_NarratorKnightley_Voice_1.wav"
prompt = "Can you generate a 200 word essay based on this audio"

# Run the async processing pipeline
nest_asyncio.apply()
asyncio.run(process_and_play(audio_file_name, prompt))
```
